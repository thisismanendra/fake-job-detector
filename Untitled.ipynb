```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

PROJECT_DIR = Path.cwd()
DATA_PATH = PROJECT_DIR / "data" / "sample_jobs.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded:", df.shape)
```

```python
df.info()
```

```python
display(df.describe(include="all"))
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate records:", df.duplicated().sum())
```

```python
print(df["fraudulent"].value_counts())

print("\nPercentage distribution:")
print(df["fraudulent"].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(8, 5))

df["fraudulent"].value_counts().sort_index().plot(
    kind="bar"
)

plt.xticks([0, 1], ["Genuine", "Fake"], rotation=0)
plt.xlabel("Job Category")
plt.ylabel("Number of Job Postings")
plt.title("Genuine vs Fake Job Distribution")
plt.tight_layout()
plt.show()
```

```python
TEXT_COLUMNS = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]

df[TEXT_COLUMNS] = df[TEXT_COLUMNS].fillna("")

print("Missing text values handled!")
```

```python
before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print("Records before:", before)
print("Records after:", after)
print("Duplicates removed:", before - after)
```

```python
print(df["fraudulent"].value_counts())

print("Unique target values:", df["fraudulent"].unique())
```

```python
df["text"] = df[TEXT_COLUMNS].agg(" ".join, axis=1)

display(df[["text", "fraudulent"]].head())
```

```python
from sklearn.model_selection import train_test_split

X = df["text"]
y = df["fraudulent"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
print("Current dataset shape:", df.shape)

print("\nGenuine and Fake distribution:")
print(df["fraudulent"].value_counts())

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
print("Dataset before duplicate removal:", before)
print("Dataset after duplicate removal:", after)
print("Duplicates removed:", before - after)
```

```python
print("Training distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

```python
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "-m", "src.train"],
    cwd=str(PROJECT_DIR),
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
else:
    print("Model training completed!")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

print("ML libraries imported successfully!")
```

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.9,
        sublinear_tf=True
    )),

    ("clf", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        C=2.0,
        random_state=42
    ))
])

print("TF-IDF + Logistic Regression pipeline created!")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1],
    target_names=["Genuine", "Fake"],
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Genuine", "Fake"]
).plot()

plt.title("Fake Job Detector - Confusion Matrix")
plt.show()
```

```python
import joblib

MODEL_DIR = PROJECT_DIR / "models"
MODEL_DIR.mkdir(exist_ok=True)

MODEL_PATH = MODEL_DIR / "fake_job_model.joblib"

joblib.dump(model, MODEL_PATH)

print("Model saved successfully!")
print("Saved location:", MODEL_PATH)
```

```python
train_accuracy = model.score(X_train, y_train)
test_accuracy = model.score(X_test, y_test)

print("Training Accuracy:", round(train_accuracy * 100, 2), "%")
print("Testing Accuracy:", round(test_accuracy * 100, 2), "%")
```

```python
from sklearn.metrics import roc_auc_score

y_probability = model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(y_test, y_probability)

print("ROC-AUC Score:", round(roc_auc, 4))
```

```python
train_texts = set(X_train.astype(str))
test_texts = set(X_test.astype(str))

overlap = train_texts.intersection(test_texts)

print("Identical texts in both sets:", len(overlap))
```
